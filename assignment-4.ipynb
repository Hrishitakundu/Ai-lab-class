```python
import networkx as nx
import matplotlib.pyplot as plt

# Graph Definition
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': ['G'],
    'G': []
}
def dfs(graph, node, visited=None, traversal=None):
    if visited is None:
        visited = set()
    if traversal is None:
        traversal = []

    visited.add(node)
    traversal.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited, traversal)

    return traversal

# Perform DFS
result = dfs(graph, 'A')
print("DFS Traversal:", " -> ".join(result))

# Create NetworkX Directed Graph
G = nx.DiGraph(graph)

plt.figure(figsize=(8, 6))

# Node Positions
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}

# Draw Graph
nx.draw_networkx_nodes(
    G,
    pos,
    node_size=1200,
    node_color="orange",
    edgecolors="black"
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=12,
    font_weight="bold"
)

nx.draw_networkx_edges(
    G,
    pos,
    arrows=True,
    arrowstyle="-|>",
    arrowsize=20,
    edge_color="black",
    width=2
)

plt.title("Graphical Representation of DFS", fontsize=14)
plt.axis("off")
plt.show()
```
